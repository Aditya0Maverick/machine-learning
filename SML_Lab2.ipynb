# Part 2 — NumPy: Numerical Computing

## Why NumPy?

The reference material emphasizes an important idea: numerical data can be represented using NumPy arrays, making mathematical operations concise and efficient.

For example, calculating a weighted score with Python lists requires explicit element-by-element work. NumPy allows the same idea to be expressed directly using array operations.

**Key idea:**

> Python lists are general-purpose containers; NumPy arrays are designed for efficient numerical computation.


## 2.1 From Python List to NumPy Array

### Concept
A NumPy array is a multidimensional structure designed for numerical computation.

### Why use it?
NumPy provides:
- convenient mathematical operations;
- multidimensional arrays;
- aggregation functions;
- broadcasting;
- vectorized computation;
- matrix operations.


```python
python_marks = [72, 81, 69, 94, 86]
numpy_marks = np.array([72, 81, 69, 94, 86])

print("Python list:", python_marks)
print("NumPy array:", numpy_marks)
print("Array type:", type(numpy_marks))
```

```python
# Compare a simple numerical operation
marks = np.array([72, 81, 69, 94, 86])

print("Marks + 5:")
print(marks + 5)
```

### Understand the output

With a NumPy array, `marks + 5` adds 5 to **every element**.

This is an example of **vectorized computation**: we express the operation on the whole array instead of writing a Python loop for each element.


## 2.2 Creating NumPy Arrays

NumPy provides several array-creation functions.

Common examples:
- `np.array()`
- `np.zeros()`
- `np.ones()`
- `np.full()`
- `np.arange()`
- `np.linspace()`
- `np.eye()`


```python
print("array:", np.array([1, 2, 3, 4]))
print("zeros:\n", np.zeros((2, 3)))
print("ones:\n", np.ones((2, 3)))
print("full:\n", np.full((2, 3), 7))
print("arange:", np.arange(2, 12, 2))
print("linspace:", np.linspace(0, 1, 5))
print("identity:\n", np.eye(3))
```

## 2.3 Array Shape, Dimensions and Data Type

### Concept
The **shape** tells us the size of the array along each dimension.

For example:
- `(5,)` → 5 values in one dimension
- `(3, 4)` → 3 rows and 4 columns
- `(2, 3, 4)` → a 3-D array

The reference material stresses that understanding shape is important when working with multidimensional data.


```python
scores = np.array([
    [72, 81, 69],
    [94, 86, 91],
    [78, 83, 88]
])

print("Shape:", scores.shape)
print("Dimensions:", scores.ndim)
print("Number of elements:", scores.size)
print("Data type:", scores.dtype)
```

## 2.4 Indexing and Slicing NumPy Arrays

NumPy extends Python's indexing idea to multiple dimensions.

For a 2-D array:

`array[row, column]`


```python
scores = np.array([
    [72, 81, 69],
    [94, 86, 91],
    [78, 83, 88]
])

print("First row:", scores[0])
print("Element at row 2, column 3:", scores[1, 2])
print("First two rows:\n", scores[:2])
print("Second column:", scores[:, 1])
```

## 2.5 Arithmetic Operations

NumPy supports element-wise operations such as `+`, `-`, `*`, `/` and `%`.

This is different from matrix multiplication. The `*` operator performs element-wise multiplication, while `@` performs matrix multiplication.


```python
a = np.array([4, 7, 10, 13])
b = np.array([2, 5, 8, 11])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Element-wise multiplication:", a * b)
print("Division:", a / b)
```

## 2.6 Aggregation and Statistics

NumPy provides functions and methods such as:
`sum`, `mean`, `median`, `std`, `min`, `max`.

For a multidimensional array, `axis` specifies the dimension along which an operation is performed.


```python
scores = np.array([
    [72, 81, 69],
    [94, 86, 91],
    [78, 83, 88]
])

print("Overall mean:", scores.mean())
print("Overall maximum:", scores.max())
print("Overall minimum:", scores.min())
print("Column means:", scores.mean(axis=0))
print("Row means:", scores.mean(axis=1))
```

## 2.7 Dot Product and Matrix Multiplication

### Concept
The dot product combines corresponding values and sums the results.

For example, if marks have different weights, a weighted score can be calculated using a dot product.

This idea in the reference notebook leads naturally from Python lists to NumPy arrays and matrix multiplication.


```python
marks = np.array([80, 70, 90])
weights = np.array([0.3, 0.2, 0.5])

weighted_score = np.dot(marks, weights)

print("Weighted score:", weighted_score)
print("Using @:", marks @ weights)
```

```python
# Matrix multiplication
student_marks = np.array([
    [80, 70, 90],
    [75, 85, 88],
    [92, 81, 79]
])

weights = np.array([0.3, 0.2, 0.5])

weighted_scores = student_marks @ weights
print("Weighted score for each student:", weighted_scores)
```

## 2.8 Reshaping and Transpose

### Concept
`reshape()` changes the organization of an array without changing the values.

`T` gives the transpose, exchanging rows and columns for a 2-D array.


```python
values = np.arange(1, 13)

matrix = values.reshape(3, 4)

print("Original:", values)
print("3 × 4 matrix:\n", matrix)
print("Transpose:\n", matrix.T)
```

## 2.9 Boolean and Comparison Operations

Comparison operators such as `==`, `!=`, `>`, `<` and `>=` return Boolean arrays.

This is useful for selecting values that satisfy a condition.


```python
marks = np.array([58, 73, 91, 67, 84])

print("Marks >= 70:", marks >= 70)
print("Selected marks:", marks[marks >= 70])
print("Marks between 60 and 90:",
      marks[(marks >= 60) & (marks <= 90)])
```

## 2.10 Broadcasting

### Concept
Broadcasting allows NumPy to perform arithmetic between arrays with compatible shapes without explicitly creating repeated copies.

Example: adding one bonus value to every subject mark, or adding a different bonus to each column.


```python
marks = np.array([
    [72, 81, 69],
    [94, 86, 91],
    [78, 83, 88]
])

bonus = np.array([3, 2, 4])

adjusted = marks + bonus

print("Original:\n", marks)
print("Bonus:", bonus)
print("Adjusted:\n", adjusted)
```

### Why did this work?

`marks` has shape `(3, 3)` and `bonus` has shape `(3,)`.

NumPy aligns the compatible dimensions and applies the bonus across each row.

If the shapes are incompatible, NumPy raises a broadcasting error.


## 2.11 Vectorization

### Concept
Vectorization means expressing an operation on an entire array rather than processing individual elements with Python loops.

For numerical workloads, vectorized NumPy operations are generally more concise and can be much faster because the underlying operations are optimized.


```python
# Loop-based calculation
values = list(range(1, 11))
loop_result = []

for value in values:
    loop_result.append(value ** 2)

# Vectorized calculation
array_values = np.arange(1, 11)
vectorized_result = array_values ** 2

print("Loop result:", loop_result)
print("Vectorized result:", vectorized_result)
```

The reference NumPy notebook uses climate variables such as **temperature, rainfall and humidity** and demonstrates weighted prediction using matrix multiplication.

Here we use a small, self-contained dataset so the lab does not depend on downloading an external file.


```python
climate_data = np.array([
    [25.0, 76.0, 99.0],
    [39.0, 65.0, 70.0],
    [59.0, 45.0, 77.0],
    [84.0, 63.0, 38.0],
    [66.0, 50.0, 52.0]
])

weights = np.array([0.3, 0.2, 0.5])

predicted_yield = climate_data @ weights

print("Climate data:\n", climate_data)
print("Predicted values:", predicted_yield)
```

## Why Pandas?

NumPy is excellent for numerical arrays. However, real datasets usually contain **rows, columns, names, categories and mixed data types**.

Pandas provides:
- Series;
- DataFrames;
- easy row/column selection;
- filtering and sorting;

## 3.1 Pandas Series

### Concept
A Series is a one-dimensional labeled data structure.

Think of it as a single column of data with an index.


```python
marks = pd.Series(
    [72, 81, 69, 94, 86],
    index=["Python", "NumPy", "Pandas", "ML", "Statistics"]
)

print(marks)
print("\nMean:", marks.mean())
```

## 3.2 Pandas DataFrame

### Concept
A DataFrame is a two-dimensional labeled table containing rows and columns.

A DataFrame is one of the most commonly used Pandas structures for data analysis.


```python
df = pd.DataFrame({
    "Student": ["Aarav", "Diya", "Kabir", "Meera", "Rohan", "Tara"],
    "Python": [78, 91, 67, 84, 73, 88],
    "NumPy": [82, 87, 71, 79, 95, 76],
    "Pandas": [74, 93, 69, 88, 81, 90],
    "Attendance": [86, 92, 79, 88, 83, 95]
})

display(df)
```

### Understand the DataFrame

Here:
- each **row** represents one student;
- each **column** represents one attribute;
- `Student` is a text column;
- subject and attendance columns contain numerical data.

This tabular structure makes the dataset convenient to inspect and analyze.


## 3.3 Inspecting Data

Before analyzing a dataset, inspect its structure.

Important commands:
- `head()` — first rows
- `tail()` — last rows
- `shape` — rows and columns
- `columns` — column names
- `dtypes` — data types
- `describe()` — numerical summary


```python
print("Shape:", df.shape)
print("\nColumns:", list(df.columns))
print("\nData types:")
print(df.dtypes)

print("\nFirst 3 rows:")
display(df.head(3))

print("\nLast 2 rows:")
display(df.tail(2))
```

```python
display(df.describe())
```

## 3.4 Selecting Columns

You can select one or multiple columns from a DataFrame.


```python
print("Student names:")
display(df["Student"])

print("Student and Pandas marks:")
display(df[["Student", "Pandas"]])
```

## 3.5 Filtering Rows

### Concept
Filtering means selecting only records that satisfy a condition.

For example:
`df[df["Attendance"] >= 85]`

This is the Pandas equivalent of asking a question such as:

> Which students have attendance of at least 85%?


```python
high_attendance = df[df["Attendance"] >= 85]
display(high_attendance)
```

```python
# Multiple conditions
selected = df[(df["Attendance"] >= 85) & (df["Pandas"] >= 80)]
display(selected)
```

## 3.6 Sorting

Sorting helps us arrange records according to a variable.

For example, we can rank students from highest to lowest Pandas marks.


```python
sorted_df = df.sort_values("Pandas", ascending=False)
display(sorted_df)
```

## 3.7 Creating Calculated Columns

A major advantage of Pandas is that we can create new columns from existing columns.

Here we calculate the average of the three subjects.


```python
df["Average_Marks"] = df[["Python", "NumPy", "Pandas"]].mean(axis=1)

display(df[["Student", "Python", "NumPy", "Pandas", "Average_Marks"]])
```

## 3.8 Conditional Categories

We can convert numerical values into meaningful categories.

For example:
- average ≥ 85 → Excellent
- average ≥ 70 → Good
- otherwise → Needs Improvement


```python
conditions = [
    df["Average_Marks"] >= 85,
    df["Average_Marks"] >= 70
]

choices = ["Excellent", "Good"]

df["Performance"] = np.select(
    conditions,
    choices,
    default="Needs Improvement"
)

display(df[["Student", "Average_Marks", "Performance"]])
```

## 3.9 Descriptive Statistics

Pandas provides convenient statistical summaries.

We can calculate the mean, minimum, maximum and standard deviation of columns.


```python
print("Average subject marks:")
print(df[["Python", "NumPy", "Pandas"]].mean())

print("\nHighest marks:")
print(df[["Python", "NumPy", "Pandas"]].max())

print("\nLowest marks:")
print(df[["Python", "NumPy", "Pandas"]].min())
```

## 3.10 Grouping and Aggregation

### Concept
`groupby()` allows us to divide records into groups and calculate statistics for each group.

For example, we can compare the average marks of different performance categories.


```python
summary = df.groupby("Performance")["Average_Marks"].agg(
    ["count", "mean", "min", "max"]
)

display(summary)
```

## 4.1 Missing Values

A missing value means that a data field has no recorded value.

`isna()` identifies missing values and `sum()` counts them.

## 4.2 Duplicate Rows

### Concept
A duplicate row is a record that is repeated exactly.

Use:
`df.duplicated()`

To count duplicates:
`df.duplicated().sum()`


```python
duplicate_rows = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_rows)

assert duplicate_rows == 0
print("✓ No duplicate rows in the main lab dataset.")
```

## 4.3 Duplicate Values in a Key Column

A dataset may have no completely duplicated rows but still contain duplicate values in an important column such as Student ID.

Use `nunique()` to compare the number of unique values with the number of records.


```python
total_records = len(df)
unique_students = df["Student"].nunique()

print("Total records:", total_records)
print("Unique student names:", unique_students)

assert total_records == unique_students
print("✓ Student names are unique.")
```

### Important distinction

**Duplicate row:** the entire record is repeated.

**Duplicate value:** a value in a particular column is repeated.

Both checks can be important depending on the dataset.

# Part 5 — NumPy + Pandas Integration

NumPy and Pandas are complementary.

**Pandas:** organizes and manages tabular data.

**NumPy:** performs efficient numerical computation.

A common workflow is:
`DataFrame → NumPy calculation → new DataFrame column`


```python
# Use NumPy to calculate a z-score for student averages
mean_score = df["Average_Marks"].mean()
std_score = df["Average_Marks"].std()

df["Z_Score"] = (df["Average_Marks"] - mean_score) / std_score

display(df[["Student", "Average_Marks", "Z_Score"]])
```

```python
# Ranking students
df["Rank"] = df["Average_Marks"].rank(
    ascending=False,
    method="min"
).astype(int)

display(
    df.sort_values("Rank")[
        ["Rank", "Student", "Average_Marks", "Performance"]
    ]
)
```

# Part 6 — Matplotlib: Data Visualization

## Why Visualization?

A table can contain many values, but a graph can make patterns easier to see.

Matplotlib is a plotting library. The reference notebook introduces:
- plotting;

## 6.1 Before Plotting: Ask the Question

Do not create a graph just because you can.

First ask:

**What do I want to understand?**

Examples:

- Observe change over time → **line plot**
- Examine relationship between two variables → **scatter plot**
- Understand distribution → **histogram**


## 6.2 Line Plot

### Use when
You want to show a trend or change across an ordered sequence, such as time or test number.


```python
test_number = np.arange(1, 7)
average_score = np.array([58, 64, 71, 69, 78, 85])

plt.figure(figsize=(7, 4))
plt.plot(test_number, average_score, marker="o")
plt.xlabel("Test Number")
plt.ylabel("Average Score")
plt.title("Average Score Across Tests")
plt.grid(True, alpha=0.3)
plt.show()
```

### Understand the graph

The x-axis shows the test number and the y-axis shows the average score.

The line helps us see the overall trend across tests.


## 6.3 Bar Chart

### Use when
You want to compare separate categories.

Here we compare the average marks of students.


```python
plt.figure(figsize=(8, 4))
plt.bar(df["Student"], df["Average_Marks"])
plt.xlabel("Student")
plt.ylabel("Average Marks")
plt.title("Student Average Marks")
plt.xticks(rotation=30)
plt.show()
```

## 6.4 Scatter Plot

### Use when
You want to investigate a possible relationship between two numerical variables.

Question:

> Do students with higher attendance tend to have higher average marks?


```python
plt.figure(figsize=(7, 4))
plt.scatter(df["Attendance"], df["Average_Marks"])
plt.xlabel("Attendance (%)")
plt.ylabel("Average Marks")
plt.title("Attendance vs Average Marks")
plt.grid(True, alpha=0.3)
plt.show()
```

### Interpret the graph

Look for:
- upward or downward patterns;
- clusters;
- unusual observations;
- whether the relationship appears strong or weak.

Remember: a scatter plot can show association, but it does not by itself prove causation.


## 6.5 Histogram

### Use when
You want to understand the **distribution** of a numerical variable.

A histogram groups values into ranges called bins.


```python
plt.figure(figsize=(7, 4))
plt.hist(df["Average_Marks"], bins=5, edgecolor="black")
plt.xlabel("Average Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Average Marks")
plt.show()
```

## 6.6 Subplots

### Concept
Subplots allow multiple related charts to be displayed in one figure.

This is useful when comparing different views of the same dataset.


```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].bar(df["Student"], df["Pandas"])
axes[0].set_title("Pandas Marks")
axes[0].set_xlabel("Student")
axes[0].set_ylabel("Marks")
axes[0].tick_params(axis="x", rotation=30)

axes[1].scatter(df["Attendance"], df["Average_Marks"])
axes[1].set_title("Attendance vs Average")
axes[1].set_xlabel("Attendance (%)")
axes[1].set_ylabel("Average Marks")

plt.tight_layout()
plt.show()
```

## 6.7 Images as NumPy Arrays — Concept

In computer vision, an image can be represented as a NumPy array.

A color image commonly has shape:

`(height, width, 3)`

where the three channels represent RGB.

This is an important connection between **NumPy arrays and visualization**.


```python
# A tiny RGB image represented as a NumPy array
image = np.zeros((60, 80, 3), dtype=np.uint8)

# Create a simple bright square
image[15:45, 25:55, :] = [255, 255, 255]

print("Image shape:", image.shape)
print("Image dimensions:", image.ndim)

plt.figure(figsize=(5, 4))
plt.imshow(image)
plt.axis("off")
plt.title("A Simple Image Stored as a NumPy Array")
plt.show()
```
